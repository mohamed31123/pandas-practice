```
import sys; sys.path.append('..')
import pandas as pd
df = pd.read_csv('../data/sales.csv', parse_dates=['date'])
df.head()
```

```
df.info()
df.describe()
```

```
df.isna().sum()
```

```
df['quantity'] = df['quantity'].fillna(df['quantity'].median()).astype(int)
df['total'] = df['quantity'] * df['unit_price']
df.head()
```

```
df.groupby('city')['total'].sum().sort_values(ascending=False)
```

```
df.pivot_table(index='city', columns='category', values='total', aggfunc='sum', fill_value=0)
```
